# 🏘️ Notebook 2: Poverty & Inequality in Ghana

> **Objective:** Has Ghana's economic growth translated into poverty reduction — and for whom? What does inequality look like beneath the headline numbers?

**Data source:** World Bank Open Data (`data/raw/poverty_indicators.csv`)

---

## 1. Setup

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys, warnings
sys.path.append('../src')
from plot_utils import apply_style, save_fig, format_pct, PALETTE, GHANA_EVENTS

warnings.filterwarnings('ignore')
apply_style()

df = pd.read_csv('../data/raw/poverty_indicators.csv')
ghana = df[df['country'] == 'Ghana'].sort_values('date').reset_index(drop=True)
print(f'Loaded: {len(ghana)} Ghana records')
```

## 2. Chart: Poverty Headcount Trend (Two Thresholds)

```python
fig, ax = plt.subplots(figsize=(13, 5))

ax.plot(ghana['date'], ghana['poverty_headcount_190'], color=PALETTE[1], linewidth=2.5,
        marker='o', markersize=5, label='Below $1.90/day (extreme poverty)')
ax.plot(ghana['date'], ghana['poverty_headcount_320'], color=PALETTE[2], linewidth=2.5,
        marker='s', markersize=5, linestyle='--', label='Below $3.20/day (lower-middle income)')

ax.fill_between(ghana['date'], ghana['poverty_headcount_190'], alpha=0.08, color=PALETTE[1])

# Annotate oil boom
ax.axvline(2011, color='#9CA3AF', linestyle=':', linewidth=1)
ax.text(2011.2, ghana['poverty_headcount_320'].max() * 0.85, 'Oil boom\n(2011)',
        fontsize=8.5, color='#9CA3AF')

format_pct(ax)
ax.set_title('Ghana Poverty Headcount Ratio (2000–2023)', fontsize=13, fontweight='bold')
ax.set_xlabel('Year')
ax.set_ylabel('Population Below Threshold (%)')
ax.legend()
sns.despine()
plt.tight_layout()
save_fig(fig, '02_poverty_headcount_trend.png')
plt.show()
```

## 3. Chart: Gini Coefficient Over Time

```python
gini = ghana.dropna(subset=['gini_index'])

fig, ax = plt.subplots(figsize=(12, 4))
ax.plot(gini['date'], gini['gini_index'], color=PALETTE[0], linewidth=2.5, marker='D', markersize=6)
ax.fill_between(gini['date'], gini['gini_index'], alpha=0.08, color=PALETTE[0])
ax.axhline(0.40, color=PALETTE[1], linestyle='--', linewidth=1.2,
           label='0.40 — common inequality threshold')

ax.set_title('Ghana Gini Coefficient (Inequality Index)', fontsize=13, fontweight='bold')
ax.set_xlabel('Year')
ax.set_ylabel('Gini Index (0 = perfect equality, 1 = perfect inequality)')
ax.set_ylim(0.30, 0.55)
ax.legend()
sns.despine()
plt.tight_layout()
save_fig(fig, '02_gini_coefficient.png')
plt.show()

print(f'\nLatest Gini: {gini["gini_index"].iloc[-1]:.3f} ({gini["date"].iloc[-1]})')
print(f'Change since 2000: {gini["gini_index"].iloc[-1] - gini["gini_index"].iloc[0]:+.3f}')
```

## 4. Chart: Human Development — Education & Health

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# School enrolment
enrol = ghana.dropna(subset=['primary_school_enrolment', 'secondary_school_enrolment'])
axes[0].plot(enrol['date'], enrol['primary_school_enrolment'], color=PALETTE[0],
             linewidth=2, label='Primary', marker='o', markersize=4)
axes[0].plot(enrol['date'], enrol['secondary_school_enrolment'], color=PALETTE[1],
             linewidth=2, label='Secondary', marker='s', markersize=4)
axes[0].set_title('School Enrolment Rate', fontweight='bold')
axes[0].set_ylabel('Gross Enrolment (%)')
axes[0].legend()
sns.despine(ax=axes[0])

# Infant mortality
infant = ghana.dropna(subset=['infant_mortality_per_1000'])
axes[1].plot(infant['date'], infant['infant_mortality_per_1000'], color=PALETTE[1],
             linewidth=2.5)
axes[1].fill_between(infant['date'], infant['infant_mortality_per_1000'], alpha=0.08, color=PALETTE[1])
axes[1].set_title('Infant Mortality Rate', fontweight='bold')
axes[1].set_ylabel('Deaths per 1,000 Live Births')
sns.despine(ax=axes[1])

plt.suptitle('Human Development Indicators: Ghana (2000–2023)', fontsize=13, fontweight='bold', y=1.02)
plt.tight_layout()
save_fig(fig, '02_human_development.png')
plt.show()
```

## 5. Key Findings — Poverty & Inequality
